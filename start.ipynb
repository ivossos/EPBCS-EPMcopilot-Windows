# EPBCS EPM Copilot - Fresh Start
Configure your environment variables below, then run the cells in order.

```python
import os

os.environ['EPBCS_VENDAS_URL']  = 'https://your-instance.pbcs.oraclecloud.com'
os.environ['EPBCS_VENDAS_USER'] = 'your_username'
os.environ['EPBCS_VENDAS_PASS'] = 'your_password'
```

```python
import requests
from requests.auth import HTTPBasicAuth

base_url  = os.environ['EPBCS_VENDAS_URL']
auth      = HTTPBasicAuth(os.environ['EPBCS_VENDAS_USER'], os.environ['EPBCS_VENDAS_PASS'])
headers   = {'Content-Type': 'application/json', 'Accept': 'application/json'}

def get(path):
    r = requests.get(f'{base_url}/HyperionPlanning/rest/v3/{path}', auth=auth, headers=headers)
    r.raise_for_status()
    return r.json()

def post(path, payload):
    r = requests.post(f'{base_url}/HyperionPlanning/rest/v3/{path}', auth=auth, headers=headers, json=payload)
    r.raise_for_status()
    return r.json()

print('Ready.')
```
